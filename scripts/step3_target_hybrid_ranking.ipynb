```python
import pickle
```

## Load all data

```python
def load_data(dataset, method):
    filename_merged_data = f"../data/intermediate/step2/{dataset}_{method}_merged_data.pkl"
    filename_processor = f"../data/intermediate/step2/{method}_{dataset}_processor.pkl"
    with open(filename_merged_data, "rb") as file:
        merged_data = pickle.load(file)
    with open(filename_processor, "rb") as file:
        processor = pickle.load(file)
    return merged_data, processor
```

## Target ranking

```python
from helpers.step2 import determine_results_traditional
from helpers.step3 import get_all_position_matches_and_sorted_lists, get_hybrid_position_match, top_k_results
```

### **Embeddings**

**NEVO**

```python
# Load data
nevo_openai_merged_data, openai_nevo_processor = load_data("nevo", "openai")
# Results on reranked form
all_sorted_lists, all_position_matches = get_all_position_matches_and_sorted_lists(nevo_openai_merged_data, openai_nevo_processor)
# Results on hybrid form
best_match_plus_proposed, hybrid_position_match = get_hybrid_position_match(nevo_openai_merged_data, openai_nevo_processor, all_sorted_lists)

# Add to df
nevo_openai_merged_data['position_match'] = all_position_matches
nevo_openai_merged_data['best_match_plus_proposed'] = best_match_plus_proposed
nevo_openai_merged_data['hybrid_position_match'] = hybrid_position_match
```

```python
# Original results
determine_results_traditional(nevo_openai_merged_data)#, labels = ['propose'])
```

```python
# Target ranking results
top_k_results(nevo_openai_merged_data, position_column='position_match')
```

```python
# Hybrid results
top_k_results(nevo_openai_merged_data, position_column='hybrid_position_match')#, labels = ['propose'])
```
